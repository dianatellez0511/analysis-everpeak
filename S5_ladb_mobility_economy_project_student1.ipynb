<details>
<summary>Haz clic para ver la pista</summary>

## 🧩Paso 4: Analizar y resumir datos de movilidad

Como el dataset de tráfico contiene **múltiples registros por ciudad**. En esta parte, calcularás los promedios anuales por ciudad para simplificar el análisis y obtener una visión más clara de las tendencias generales.

### 4.1 Calcular promedios de tráfico por ciudad

**🎯Objetivo:**
Obtener una vista consolidada del tráfico promedio por ciudad y año, para analizar patrones generales sin depender de datos diarios.

**Instrucciones**

- Agrupa los datos por `city`, `country` y `year`.
- Calcula el promedio **solo de las métricas de tráfico más relevantes**: como `jams_delay`, `traffic_index_live`, `jams_length_kms`, `jams_count`, `mins_delay`, y tiempos de viaje (`travel_time_live_per_10kms_mins` y `travel_time_hist_per_10kms_mins`).
- Guarda el resultado como `traffic_city_year_2024`, mantén las columnas como variables (no índices).


<details>
<summary>Haz clic para ver la pista</summary>
Usa ".agg()" para aplicar funciones de promedio. Al final, reinicia el índice para mantener las columnas de la agrupación como variables (no índices).

```python
# Calcular los  promedios de trafico por ciudad, país y año
traffic_city_year_2024 =  traffic_2024.groupby(['city', 'country', 'year']).agg({'jams_delay':'mean',
 'traffic_index_live':'mean','jams_lenghtin_kms':'mean','jams_count':'mean','mins_delay':'mean','travel_time_live_per_10kms_mins':'mean',
'travel_time_historic_per_10Kms_mins':'mean'}).reset_index()

# Mostrar resultado
traffic_city_year_2024.head()
```

## 🧩Paso 5: Unir movilidad y economía

Combinar datasets te permite analizar cómo se relacionan los indicadores económicos con los de movilidad.

### 5.1 Unir tráfico (tabla principal) con indicadores económicos

**🎯Objetivo:**
Combinar la información de tráfico y economía en un solo DataFrame para analizar cómo las condiciones económicas se relacionan con la movilidad urbana.

**Instrucciones**
- Selecciona solo las **columnas relevantes** de cada dataset (por ejemplo, variables clave de tráfico y de economía).
- Usa `.copy()` al crear subconjuntos para evitar modificar el dataset original.
- Une ambos DataFrames y define como **claves de unión** a `city` y `year`.
- Mantén solo las ciudades y años presentes en ambos datasets.
- Guarda el resultado en una nueva variable llamada `merged` y muestra las primeras 5 filas.


<details>
<summary>Haz clic para ver la pista</summary>
Aplica una unión de tipo "inner" para mantener las ciudades y años presentes en ambos datasets.

```python

# Seleccionar columnas clave de tráfico y economía
left_cols = ['city','country','year','jams_delay','traffic_index_live',
             'jams_lenghtin_kms','jams_count','mins_delay',
             'travel_time_live_per_10kms_mins','travel_time_historic_per_10Kms_mins']

right_cols = ['city','year','city_gdp_capita','unemployment_pct','pm25','population']

# Usar .copy() para crear los dos nuevos datasets reducidos
# traffic_2024_small = traffic_city_year_2024[ ... ].copy()
traffic_2024_small=traffic_city_year_2024[left_cols].copy()
# eco_2024_small = ...
eco_2024_small=eco_2024[right_cols].copy()

# Unir datasets
merged = traffic_2024_small.merge(eco_2024_small,on=['city','year'],how='inner')

# Mostrar las primeras 5 filas
# tu código aquí
merged.head(5)


```

# 🧾 Resumen ejecutivo



**Contexto & objetivo:**  
El presente estudio analiza la relación que existe entre movilidad urbana y productividad económica de las principales ciudades de Latinoamerica. Su propósito es proporcionar una base análitica al Banco de Desarrollo para identificar zonas estratégicas de inversión en movilidad urbana. Para abordar este tema, se examinan dos variables clave, la congestión vial que mide la saturación de la red de transporte a través de los tiempos de retraso y el PIB per cápita, que representa el ingreso económico por habitante. Al evaluar ambos indicadores en conjunto, el banco cuenta con información clave para decidir dónde es más conveniente invertir en infraestructura

**Cobertura de datos:**  
Para este estudio se evaluaron 15 ciudades de 7 países latinoamericanos (Argentina, Brasil, Chile, Colombia, México, Perú y Uruguay), con información registrada durante el año 2024.

**Metodología:**

Para la metodología del estudio inicialmente se estandarizó la estructura de las tablas unificando los nombres de las columnas, corrigiendo también los formatos de fecha y texto a valores númericos y se removieron los carácteres especiales como: símbolos de porcentaje y separadores décimales garantizando con esto una base uniforme. Posteriormente se promediaron los registros de la base de datos para obtener un único valor anual por ciudad durante el año 2024 y esta información de movilidad se cruzo con los indicadores económicos de las mismas ciudades, conservando únicamente aquellos con datos disponibles en ambas fuentes. Finalmente, se ejecutarón validaciones visuales para inspeccionar la distribución de las variables, evaluar el comportamientos de los datos e identificar valores extremos desproporcionados antes de realizar el análisis comparativo definitivo.  

**Hallazgos iniciales:**  

El análisis realizado sobre el DataFrame integrado (merged) permitió determinar la relación entre el producto interno bruto per cápita (city_gdp_capita) y los tiempos de retraso por tráfico (jams_delay) en 15 ciudades de la región.

Se encontró que no existe una correlación lineal entre el PIB per cápita y la congestión vehicular. Esta independencia se evidencia en ciudades con alto PIB per cápita y baja congestión vehicular como Montevideo y Brasilia: Montevideo con un PIB per cápita de 26176 y Brasilia con 16251. Por su parte, los niveles de congestión en Montevideo son de 50,20 y en Brasilia de 101,57.

Por otro lado, tenemos ciudades con PIB per cápita medio y un alto nivel de congestión vehicular, como Bogotá y Lima: Bogotá con un PIB per cápita de 11442 y Lima con 13472. En contraste con esto, los niveles de congestión en Bogotá son de 1141,55 y en Lima de 1052,29.

Finalmente, Ciudad de México coincide con valores altos en ambas métricas, con un PIB per cápita de 21111 y una congestión vehicular de 2833,05, lo que confirma la independencia de las variables. Su alta congestión podría deberse a factores relacionados como la densidad poblacional, la limitada infraestructura vial o el exceso de medios de transporte por habitante.

**Recomendaciones**  

A partir de estos hallazgos, se evidencia que las decisiones de inversión en infraestructura de movilidad no deben guiarse por el nivel de riqueza de una ciudad, sino por el impacto directo del tráfico sobre su productividad. En consecuencia, se sugieren las siguientes acciones:

1. Priorización de inversión en infraestructura:
   * Bogotá y Lima (Prioridad Alta): Deben ser consideradas prioritarias para el financiamiento de proyectos de transporte masivo de alta capacidad.    * Ciudad de México (Enfoque especial): Aunque registra la mayor congestión del estudio, su alto PIB per cápita le otorga mayor capacidad fiscal propia. Se recomienda orientar el apoyo del banco hacia la asistencia técnica para la gestión de demanda de tráfico y optimización tecnológica, en lugar de financiamiento directo de obras básicas.
